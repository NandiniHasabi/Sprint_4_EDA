# Data Inspection


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

## 1. Loading the Dataset

### Concept
Dataset loading is the process of importing data into Python for inspection and analysis.

Pandas provides the read_csv() function to load a CSV file into a DataFrame.

### Example
An organization may store customer details, transactions, sales, or machine learning training data in a CSV file.

### Importance
Loading the dataset is the first step in understanding its structure before performing Exploratory Data Analysis or building Machine Learning models.

```python
df = pd.read_csv("ml_training_data.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

## 2. Dataset Dimensions

### Concept
Dataset dimensions describe the size of a dataset.

- Rows represent records or observations.
- Columns represent variables or features.

The shape attribute returns the number of rows and columns.

### Example
A dataset with 1,000 rows and 8 columns contains 1,000 records and 8 variables.

### AI/ML Relevance
Dataset dimensions help us understand how much data is available for analysis and model development.

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset dimensions:", df.shape)
```

## 3. Column Names

### Concept
Column names identify the variables present in a dataset.

### Example
Columns in a training dataset may represent age, salary, experience, department, or a target label.

### Importance
Inspecting column names helps us understand the available information and identify possible input features, target variables, and identifier columns.

```python
print("Column names:")
print(df.columns.tolist())

print("\nTotal number of columns:", len(df.columns))
```

## 4. Dataset Index

### Concept
The index identifies the rows in a Pandas DataFrame. By default, Pandas assigns integer labels starting from zero.

An index is different from an identifier column such as Customer_ID or Employee_ID.

### Importance
Understanding the index helps us avoid mistakenly treating row labels as meaningful predictive features.

```python
print("Dataset index:")
print(df.index)

print("\nFirst five index values:")
print(df.index[:5])
```

## 10. Using dtypes

### Concept
The dtypes attribute displays the data type of each column.

Common data types include:
- int64: integer values
- float64: decimal values
- object: often text or mixed values
- bool: Boolean values
- datetime64: date and time values

### AI/ML Relevance
Incorrect data types can cause problems during analysis and model preparation.

```python
print(df.dtypes)
```

## 11. Using nunique()

### Concept
The nunique() method counts the number of distinct values in each column. By default, missing values are excluded.

### Importance
It helps identify columns with few categories, columns with many distinct values, and possible identifier columns.

```python
print("Unique values in each column:")
print(df.nunique())
```

## 12. Using value_counts()

### Concept
The value_counts() method counts how frequently each value occurs in a column.

### Example
A dataset may contain different departments, product categories, or customer groups.

### Importance
Frequency counts help identify common categories, rare categories, and possible class imbalance.

```python
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

if categorical_columns:
    column = categorical_columns[0]

    print("Selected column:", column)
    print("\nValue counts:")
    print(df[column].value_counts(dropna=False))

    print("\nPercentage distribution:")
    print(
        (df[column].value_counts(normalize=True, dropna=False) * 100)
        .round(2)
    )
else:
    print("No text or categorical columns detected.")
```

## 13. Identifying Numerical Columns

### Concept
Numerical features represent quantities or measurable values.

Examples include age, salary, price, quantity, and sales.

Numerical features can be discrete, such as the number of products sold, or continuous, such as weight or height.

### Importance
Identifying numerical columns helps us decide which statistical calculations and visualizations to use.

```python
numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nNumber of numerical columns:", len(numerical_columns))
```

## 14. Identifying Categorical Columns

### Concept
Categorical features represent labels or groups rather than measurable quantities.

Examples include department, city, product category, and payment method.

### Importance
Categorical columns can be examined using frequency tables and bar charts. Some categorical variables may be stored as numbers, so their meanings should also be checked.

```python
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Categorical or text columns:")
print(categorical_columns)

print("\nNumber of detected categorical columns:", len(categorical_columns))
```

## 15. Identifying Date/Time Columns

### Concept
Date/time columns contain dates, times, or timestamps.

Examples include order date, joining date, and transaction timestamp.

### Importance
Date/time columns can help us identify trends, seasonal patterns, and changes over time.

```python
datetime_columns = df.select_dtypes(
    include=["datetime", "datetimetz"]
).columns.tolist()

possible_date_columns = [
    col for col in df.columns
    if any(word in col.lower() for word in ["date", "time", "timestamp"])
]

print("Columns stored as datetime:")
print(datetime_columns)

print("\nColumns with possible date/time names:")
print(possible_date_columns)
```
