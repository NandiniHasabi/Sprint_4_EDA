# Dataset Understanding

### 1. What is a Dataset?

A dataset is a collection of related data organized in rows and columns.

- Rows represent individual records or observations.
- Columns represent features or attributes.
- Each cell contains a value for a particular record and feature.

**Simple example:**

A student dataset may contain student names, ages, marks, and results.

**Real-world example:**

A bank dataset may contain customer details, account balances, and loan information.

**AI/ML use case:**

A Machine Learning model learns patterns from a dataset to make predictions.

**Why is it important?**

Understanding a dataset helps us know what information is available and whether it is suitable for analysis.


### 2. What is a Feature?

A feature is an individual column that describes an observation.

Examples:
- Age
- Fare
- Passenger class
- Gender

In a passenger dataset, age and fare are features that may help explain passenger survival.

**AI/ML relevance:**

Features are used as input variables when training a Machine Learning model. Selecting useful features can improve model performance.

### 3. What is a Target?

A target is the variable that a Machine Learning model is expected to predict.

In the Titanic dataset, `survived` is the target variable.

- `0` means the passenger did not survive.
- `1` means the passenger survived.

**Example:**

A model can use age, passenger class, and fare to predict whether a passenger survived.

**AI/ML relevance:**

Identifying the target correctly is necessary for supervised learning. The target must not accidentally be included among the input features.

### 4. Independent and Dependent Variables

**Independent variables:**

Independent variables are the input variables used to explain or predict an outcome. They are commonly represented by `X`.

Examples: age, fare, sex, and passenger class.

**Dependent variable:**

The dependent variable is the outcome we want to predict. It is commonly represented by `y`.

Example: survival status.

```python
X = df.drop(columns=["survived"])
y = df["survived"]
```

Here, `X` contains the input columns and `y` contains the target.

**AI/ML relevance:**

Separating independent variables from the dependent variable prepares the dataset for supervised Machine Learning.

### 5. Numerical Features

Numerical features contain numeric values.

Examples:
- Age
- Fare
- Number of siblings or spouses aboard (`sibsp`)

Numerical features can be used in mathematical calculations and statistical analysis.

### 6. Categorical Features

Categorical features represent groups or categories.

Examples:
- Sex
- Embarkation port (`embarked`)
- Passenger class (`class`)

Categorical features may need encoding before being used by many Machine Learning algorithms.

### 7. Discrete Features

Discrete features contain countable values, usually whole numbers.

Examples:
- Number of parents or children aboard (`parch`)
- Number of siblings or spouses aboard (`sibsp`)

### 8. Continuous Features

Continuous features can take any value within a range.

Examples:
- Age
- Fare

For example, age can be recorded as 25.5 years.

**Important:** A number does not always represent a numerical measurement. Passenger class, for example, represents an ordered category.

### 9. Identifier Columns

Identifier columns are used to identify individual records.

Examples:
- Customer ID
- Student ID
- Passenger ID

An identifier is not necessarily useful for predicting the target. A unique identifier may not contain meaningful information about the outcome.

In the Titanic dataset, `who`, `name`, and other passenger details can help describe records, but their usefulness depends on the analysis.

**AI/ML relevance:**

We should examine identifier columns carefully before including them as model features. Unique IDs may cause a model to learn irrelevant patterns.

### 10. Target Variable

A target variable is the specific column selected for prediction.

For this notebook:

- Target column: `survived`
- Target type: Binary classification
- Class `0`: Did not survive
- Class `1`: Survived

The target distribution should be analyzed to understand whether the classes are balanced.

**AI/ML relevance:**

Understanding the target helps determine the Machine Learning problem type and guides model evaluation.

### 11. Metadata

Metadata is information that describes a dataset.

It may include:
- Dataset name
- Data source
- Number of rows and columns
- Column descriptions
- Data types
- Collection method
- Missing-value information

**Example:**

The Titanic passenger dataset contains passenger details and survival information.

**Why is metadata important?**

Metadata helps us understand where the data came from, what its columns mean, and how the dataset can be used responsibly.

### 12. Dataset Dimensions

Dataset dimensions describe the number of rows and columns.

- Rows represent observations.
- Columns represent features or attributes.
- The `shape` attribute returns `(number_of_rows, number_of_columns)`.

**Example:**

If a dataset has 100 rows and 5 columns, its shape is `(100, 5)`.

**AI/ML relevance:**

Dataset dimensions help us understand the amount of available data and the number of variables that may require analysis.

### 13. Selected Dataset

**Dataset name:** Titanic Passenger Dataset

**Dataset source:** Seaborn Data Repository on GitHub

**Domain:** Transportation and passenger survival analysis

**Purpose:** To study passenger characteristics and survival outcomes.

**Target variable:** `survived`

**Why I selected this dataset:**

I selected the Titanic dataset because it contains numerical and categorical features, a binary target variable, and missing values in some columns. It is suitable for practicing dataset understanding and exploratory data analysis.

The row and column counts will be obtained directly from the loaded dataset.

````python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset name: Titanic Passenger Dataset")
print("Dataset shape:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
````

**Interpretation:**

The `head()` method displays the first five records by default. It helps me understand how the data is organized and what values appear in each column.

**Insight:**

**Interpretation:**

This code displays all column names in the dataset.

**Insight:**

Column names help me identify which variables describe passengers and which variable represents the prediction target.

````python
print("Data types of each column:")
print(df.dtypes)
````

**Interpretation:**

The `dtypes` attribute shows the data type of each column.

- `int64` represents integer values.
- `float64` represents decimal values.
- `object` often represents text or categorical values.

**Insight:**

Checking data types helps me identify how the columns are stored. I should not assume that every numeric column is a measurement or that every object column is a category without examining its meaning.

````python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical or text columns:")
print(categorical_columns)
````

**Interpretation:**

The code separates columns based on their stored data types.

**Insight:**

Numerical columns can be examined using statistics, while categorical columns can be studied using category counts and percentages. Some numerical columns may actually represent categories, so I must consider their meaning as well.

````python
column_descriptions = {
    "survived": "Survival outcome: 0 = No, 1 = Yes",
    "pclass": "Passenger class: 1, 2, or 3",
    "sex": "Passenger sex",
    "age": "Passenger age in years",
    "sibsp": "Number of siblings or spouses aboard",
    "parch": "Number of parents or children aboard",
    "fare": "Passenger fare",
    "embarked": "Port of embarkation",
    "class": "Passenger class category",
    "who": "Passenger group description",
    "adult_male": "Whether the passenger is an adult male",
    "deck": "Passenger deck",
    "embark_town": "Town associated with embarkation",
    "alive": "Survival status represented as text",
    "alone": "Whether the passenger was travelling alone"
}

for column in df.columns:
    description = column_descriptions.get(
        column, "Refer to the dataset documentation"
    )
    print(f"{column}: {description}")
````

**Interpretation:**

This code provides a short description of the important columns.

**Insight:**

Understanding the meaning of each column helps me choose suitable variables for analysis and avoid incorrect interpretations.

````python
target_column = "survived"

X = df.drop(columns=[target_column])
y = df[target_column]

print("Target variable:", target_column)
print("Input feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget values:")
print(y.value_counts(dropna=False))
````

**Interpretation:**

The input features are stored in `X`, and the target variable is stored in `y`.

The value counts show how many passengers belong to each survival class.

**Insight:**

This is a binary classification problem because the target represents two outcomes. The class counts also help identify whether the target classes are imbalanced.

**Important:** This separation is for learning purposes. Before model training, I would also review other columns, including `alive`, because it directly duplicates the survival outcome and could cause target leakage.

````python
print("Missing values by column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
````

**Interpretation:**

The code counts the missing values in each column.

**Insight:**

Missing values are a data-quality issue that should be documented. For example, age or deck information may be unavailable for some passengers.

Missing values may affect analysis and model training. In this sprint, I will identify and document the issue rather than perform complete data preprocessing.
