
## 1. What is Categorical Analysis?

Categorical analysis is used to study data that belongs to different groups or categories.

Examples include:
- Department: HR, IT, Sales
- Gender: Male, Female
- Product type: Mobile, Laptop, Tablet

It helps us understand category frequencies, proportions, and patterns in the dataset.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
import os

file_paths = [
    "ml_training_data.csv",
    r"C:\Users\giris\ml_training_data.csv",
    r"C:\Users\giris\OneDrive\Desktop\sprint-4\ml_training_data.csv"
]

file_path = next((p for p in file_paths if os.path.exists(p)), None)

if file_path is None:
    raise FileNotFoundError(
        "Dataset not found. Check the location of ml_training_data.csv."
    )

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

## 2. Identify Categorical Features

Categorical features contain labels or groups rather than continuous numerical measurements.

We will identify columns with object, category, or Boolean data types.


```python
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("Number of categorical columns:", len(categorical_columns))
```

## 3. Frequency Counts

Frequency counts show how many times each category appears in a column.

They help us identify common and uncommon categories.


```python
if categorical_columns:
    for column in categorical_columns:
        print(f"\nFrequency counts for {column}:")
        display(df[column].value_counts(dropna=False))
else:
    print("No categorical columns were found in the dataset.")
```

## 4. Category Percentages

Percentages help us compare categories even when datasets have different numbers of records.

We will calculate the percentage represented by each category.


```python
if categorical_columns:
    for column in categorical_columns:
        print(f"\nCategory percentages for {column}:")
        percentages = (
            df[column].value_counts(normalize=True, dropna=False) * 100
        ).round(2)
        display(percentages.to_frame(name="Percentage"))
else:
    print("Category percentages cannot be calculated because no categorical columns were found.")
```

## 5. Visualize Category Frequencies

Bar charts make it easier to compare the number of records in each category.


```python
if categorical_columns:
    for column in categorical_columns:
        plt.figure(figsize=(8, 4))
        order = df[column].astype("string").fillna("Missing").value_counts().index

        sns.countplot(
            x=df[column].astype("string").fillna("Missing"),
            order=order
        )

        plt.title(f"Category Frequencies: {column}")
        plt.xlabel(column)
        plt.ylabel("Count")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
else:
    print("No categorical columns are available for bar charts.")
```

## 6. Analyze Missing Values

Missing values can affect category counts and data quality.

We will check the number of missing values in each categorical feature.


```python
if categorical_columns:
    missing_counts = df[categorical_columns].isnull().sum()
    display(missing_counts.to_frame(name="Missing Values"))
else:
    print("No categorical columns are available for missing-value analysis.")
```

## 7. Cardinality Analysis

Cardinality means the number of unique categories in a feature.

A column with many unique values may need special handling during machine learning preprocessing.


```python
if categorical_columns:
    cardinality = pd.DataFrame({
        "Unique Categories": df[categorical_columns].nunique(dropna=False)
    })
    display(cardinality)
else:
    print("No categorical columns are available for cardinality analysis.")
```

## 8. Identify the Most Frequent Category

The most frequent category is the category with the highest count in a feature.

This helps us understand which groups are most common in the dataset.


```python
if categorical_columns:
    for column in categorical_columns:
        counts = df[column].value_counts(dropna=False)

        if not counts.empty:
            print(f"{column}: Most frequent category = {counts.index[0]}")
            print(f"Count = {counts.iloc[0]}")
else:
    print("No categorical columns are available to identify the most frequent category.")
```

- Calculated category percentages to understand the distribution of categories.

- Examined the number of unique categories in each feature.
- Categorical analysis helps identify common categories, missing values, and possible imbalances in the data.
